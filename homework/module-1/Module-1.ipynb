```python
# Q5: Select cars from Asia, then find their maximum fuel efficiency.
car_data.loc[car_data["origin"] == "Asia", "fuel_efficiency_mpg"].max()
```

```python
# Q6, step 1: Calculate the original median horsepower.
# The median is the middle value after sorting; Pandas skips missing values.
car_data["horsepower"].median()
```

```python
# Q6, step 2: Find the most frequent horsepower (the mode).
# mode() returns a Series; so getting first val as a number using iloc[0] its first value as a number.
most_common_hp = car_data["horsepower"].mode().iloc[0]
```

```python
# Q6, steps 3–4: Fill missing horsepower values with the mode.
# Keep the filled Series separately so the original data stays available.
# Compare this median with step 1 to determine whether it changed.
car_data_hp_imputed = car_data["horsepower"].fillna(most_common_hp)
car_data_hp_imputed.median()
```

```python
# Q7, step 1: Keep only cars whose origin is Asia, preserving row order.
asian_cars = car_data[car_data["origin"] == "Asia"]
asian_cars
```

```python
# Q7, steps 2–4: Select weight and year, in that order, for the first 7 cars.
# Convert the selected table into a NumPy array with shape (7, 2).
X = np.array(asian_cars[["vehicle_weight", "model_year"]][0:7])
```

```python
# Inspect the feature matrix: each row is a car, each column is a feature.
X
```

```python
# Q7, step 5: Transpose X by swapping rows and columns.
# XT has shape (2, 7).
XT = X.T
XT
```

```python
# Multiply the transpose by X to obtain a (2, 2) matrix.
# @ performs matrix multiplication; * would multiply element by element.
XTX = XT @ X
XTX
```

```python
# Q7, step 6: Compute the inverse of XTX for the weight calculation.
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
# Q7, step 7: Create the target values provided in the homework.
# The 7 values correspond to the 7 rows of X, in the same order.
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
# Q7, step 8: Calculate the two linear regression weights.
# The formula is w = inverse(X.T @ X) @ X.T @ y.
w = XTX_inv @ XT @ y
w
```

```python
# Q7, step 9: Sum the weights and compare with the answer choices.
w.sum()
```
